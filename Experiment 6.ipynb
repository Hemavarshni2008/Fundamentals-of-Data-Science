```python
import pandas as pd
import matplotlib.pyplot as plt

# Create website visitor data
data = {
    "Date": [
        "2026-04-01",
        "2026-04-02",
        "2026-04-03",
        "2026-04-04",
        "2026-04-05",
        "2026-04-06",
        "2026-04-07"
    ],
    "Visitors": [200, 250, 230, 280, 260, 310, 295]
}

# Create DataFrame
df = pd.DataFrame(data)

# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"])

# Set Date as index
df.set_index("Date", inplace=True)

# Display the data
print("Website Visitor Data:")
print(df)

# Display basic information
print("\nStatistical Summary:")
print(df.describe())

# Calculate average visitors
print("\nAverage Website Visitors:", df["Visitors"].mean())

# Find maximum visitors
print("Maximum Website Visitors:", df["Visitors"].max())

# Find minimum visitors
print("Minimum Website Visitors:", df["Visitors"].min())

# Plot the time series
plt.figure(figsize=(10, 5))

plt.plot(
    df.index,
    df["Visitors"],
    marker="o"
)

plt.title("Time Series Analysis of Website Visitors")
plt.xlabel("Date")
plt.ylabel("Number of Visitors")
plt.xticks(rotation=45)
plt.grid(False)
plt.tight_layout()

plt.show()
```
