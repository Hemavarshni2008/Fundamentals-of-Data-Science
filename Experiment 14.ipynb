```python
import numpy as np

import pandas as pd

from sklearn.preprocessing import StandardScaler, MinMaxScaler

# New dataset

data = {

    "Age": [19, 21, 23, 25, 27],

    "Income": [18000, 22000, 28000, 32000, 40000],

    "Experience": [1, 2, 4, 6, 8]

}

df = pd.DataFrame(data)

print("Original Dataset:")

print(df)

# Standard Scaling

sc = StandardScaler()

standard_scaled = sc.fit_transform(df)

print("\nStandard Scaled Data:")

print(standard_scaled)

# Min-Max Scaling

mms = MinMaxScaler()

minmax_scaled = mms.fit_transform(df)

print("\nMin-Max Scaled Data:")

print(minmax_scaled)
```
