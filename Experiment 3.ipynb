```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']

Product_X = [90, 130, 170, 150, 200, 240]
Product_Y = [110, 120, 155, 180, 190, 220]

plt.plot(months, Product_X, marker='o', label='Product X')
plt.plot(months, Product_Y, marker='o', label='Product Y')

plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ['Tamil', 'Biology', 'Economics', 'History', 'Computer']
marks = [82, 89, 76, 91, 96]

plt.bar(subjects, marks)

plt.title("Subject Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ['HTML', 'CSS', 'SQL', 'Java']
students = [30, 20, 25, 25]

plt.pie(students, labels=labels, autopct='%1.1f%%')

plt.title("Technology Preferences")

plt.show()
```
